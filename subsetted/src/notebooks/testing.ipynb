# ML Model Testing

Running this file requires access to all the saved models (`.pkl` and `.pth`).

## Load Data

```python
import os 

# path to features directory -- update if necessary
DATA_DIR = os.path.join('..', '..', 'data', 'features')
```

```python
import pandas as pd

test_path = os.path.join(DATA_DIR, 'test.csv')
test_df = pd.read_csv(test_path)

test_df.head()
```

## Import Models

```python
import pickle
MODELS_DIR = os.path.join('..', '..', 'models')

lr = pickle.load(open(os.path.join(MODELS_DIR, 'lr.pkl'), 'rb'))
rf = pickle.load(open(os.path.join(MODELS_DIR, 'rf.pkl'), 'rb'))
mlp = pickle.load(open(os.path.join(MODELS_DIR, 'mlp.pkl'), 'rb'))
xgb = pickle.load(open(os.path.join(MODELS_DIR, 'xgb.pkl'), 'rb'))
lgb = pickle.load(open(os.path.join(MODELS_DIR, 'lgb.pkl'), 'rb'))
```

## Prepare Testing Data

```python
# get features
feature_cols = lr.feature_names_in_
feature_cols
```

```python
X_test = test_df[feature_cols]
X_test.head()
y_test = test_df['label']
CLASSES = ['ANG', 'DIS', 'FEA', 'HAP', 'NEU', 'SAD']
classes = ['ANG', 'DIS', 'FEA', 'HAP', 'NEU', 'SAD']
class_to_num = {cls: i for i, cls in enumerate(classes)}
num_to_class = {i: cls for i, cls in enumerate(classes)}

y_test_num = [class_to_num[cls] for cls in y_test]
```

## Test each Model

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

```python
lr_preds = lr.predict(X_test)
print("Logistic Regression Accuracy: ", accuracy_score(y_test, lr_preds))
print("Logistic Regression Class-Wise Precision: ", precision_score(y_test, lr_preds, average=None))
print("Logistic Regression Class-Wise Recall: ", recall_score(y_test, lr_preds, average=None))

# random forest
rf_preds = rf.predict(X_test)
print("Random Forest Accuracy: ", accuracy_score(y_test, rf_preds))
print("Random Forest Class-Wise Precision: ", precision_score(y_test, rf_preds, average=None))
print("Random Forest Class-Wise Recall: ", recall_score(y_test, rf_preds, average=None))

# perceptron
perceptron_preds = mlp.predict(X_test)
print("Perceptron Accuracy: ", accuracy_score(y_test, perceptron_preds))
print("Perceptron Class-Wise Precision: ", precision_score(y_test, perceptron_preds, average=None))
print("Perceptron Class-Wise Recall: ", recall_score(y_test, perceptron_preds, average=None))

# xgboost
xgb_preds = xgb.predict(X_test)
print("XGBoost Accuracy: ", accuracy_score(y_test_num, xgb_preds))
print("XGBoost Class-Wise Precision: ", precision_score(y_test_num, xgb_preds, average=None))
print("XGBoost Class-Wise Recall: ", recall_score(y_test_num, xgb_preds, average=None))

# lightgbm
lgb_preds = lgb.predict(X_test).argmax(axis=1)
print("LightGBM Accuracy: ", accuracy_score(y_test_num, lgb_preds))
print("LightGBM Class-Wise Precision: ", precision_score(y_test_num, lgb_preds, average=None))
print("LightGBM Class-Wise Recall: ", recall_score(y_test_num, lgb_preds, average=None))
```

# Neural Network / ViT Testing

## Load Testing Data

```python
import torch
from PIL import Image
from tqdm import tqdm

DATA_DIR = os.path.join('..', '..', 'data', 'spectrograms')

val_transform = torch.load(os.path.join(MODELS_DIR, 'val_transform.pth'), weights_only=False)

classes = os.listdir(os.path.join(DATA_DIR, 'test'))

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

## Test CNN

```python
import torch.nn as nn

class CNN(nn.Module):
    
    def __init__(self, num_classes: int):
        super(CNN, self).__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 28 * 28, 512),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(512, num_classes)
        )
    
    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x
```

```python
cnn = CNN(num_classes=len(classes)).to(device)
cnn.load_state_dict(torch.load(os.path.join(MODELS_DIR, 'cnn.pth'), weights_only=False, map_location=device))
cnn.to(device)
cnn.eval()

preds = []
labels = []

for cls in classes:
    images = os.listdir(os.path.join(DATA_DIR, 'test', cls))
    for image in tqdm(images):
        sample_image = Image.open(os.path.join(DATA_DIR, 'test', cls, image)).convert('RGB')
        sample_image = val_transform(sample_image).unsqueeze(0).to(device)

        with torch.no_grad():
            output = cnn(sample_image)
            _, predicted = output.max(1)

        label = image.split('_')[2]
        predicted_label = classes[predicted.item()]

        preds.append(predicted_label)
        labels.append(label)
```

```python
accuracy = accuracy_score(labels, preds)
recall = recall_score(labels, preds, average=None)
precision = precision_score(labels, preds, average=None)
print(f'Accuracy: {accuracy:.2f}')
print(f'Recall: {recall}')
print(f'Precision: {precision}')
```

## Test ViT

```python
import timm

VIT_MODEL_NAME = 'vit_tiny_patch16_224'

vit = timm.create_model(
    VIT_MODEL_NAME,
    pretrained=True,
    num_classes=len(CLASSES)
).to(device)
vit.load_state_dict(torch.load(os.path.join(MODELS_DIR, 'vit.pth'), weights_only=False, map_location=device))
vit.eval()

preds = []
labels = []

for cls in classes:
    images = os.listdir(os.path.join(DATA_DIR, 'test', cls))
    for image in tqdm(images):
        sample_image = Image.open(os.path.join(DATA_DIR, 'test', cls, image)).convert('RGB')
        sample_image = val_transform(sample_image).unsqueeze(0).to(device)

        with torch.no_grad():
            output = vit(sample_image)
            _, predicted = output.max(1)

        label = image.split('_')[2]
        predicted_label = classes[predicted.item()]
        
        preds.append(predicted_label)
        labels.append(label)
```

```python
accuracy = accuracy_score(labels, preds)
recall = recall_score(labels, preds, average=None)
precision = precision_score(labels, preds, average=None)
print(f'Accuracy: {accuracy:.2f}')
print(f'Recall: {recall}')
print(f'Precision: {precision}')
```
